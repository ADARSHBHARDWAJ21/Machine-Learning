#Breast cancer claasification using logistic regression

  Import all Required Libraries

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

Load the Dataset

0 -> Benig which is safe
1 -> malignant which is not safe

```python
data = load_breast_cancer()
data
```

```python
X = pd.DataFrame(data.data , columns = data.feature_names)
y = pd.Series(data.target)
```

Binary Classification Problem
1. 0 => Malignant
2. 1 => Benign


```python
y.value_counts()
```

```python
X.head()
```

Observation: For 0's it Occupies almost 38% Whereas 1's it Occupies 62%

#Stratify - The Stratify parameter in scikit-learn's train_test_split function ensures that the traning and testing subsets maintain the exact same proportion of class labels as the original dataset

Data Split

```python
from sklearn.model_selection  import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, stratify=y)
```

```python
X_train.shape
```

```python
X_test.shape
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Model Traning: Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
```

Model Prediction

```python
y_pred = model.predict(X_test_scaled)
print(y_pred)
```

Model Predicted Probablities

Probablity of class 1 and 0 both

```python
probs_test = model.predict_proba(X_test_scaled)
print(probs_test)
```

Probablity of class 1

```python
probs_test = model.predict_proba(X_test_scaled)[: ,1]
print(probs_test)
```

Probablity for class 0

```python
probs_test = model.predict_proba(X_test_scaled)[: ,0]
print(probs_test)
```

Model Evaluation

```python
from sklearn.metrics import confusion_matrix, accuracy_score, recall_score, precision_score, f1_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
print("The overall accuracy is: ",accuracy_score(y_test, y_pred))
```

```python
print("The precision Valur is " , precision_score(y_test, y_pred))
```

```python
print("The recall value is: " , recall_score(y_test , y_pred))
```

```python
print("The f1 score value is: " ,f1_score(y_test, y_pred))
```
